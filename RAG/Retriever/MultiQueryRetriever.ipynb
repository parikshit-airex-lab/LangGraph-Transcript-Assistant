```python
from langchain_community.vectorstores import FAISS, Chroma
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_groq import ChatGroq
# from langchain_community.retrievers.multi_query import MultiQueryRetriever
from langchain_classic.retrievers import MultiQueryRetriever
```

create ambiguous documents

```python
docs = [
    Document(page_content = "Sunlight plays a crucial role in maintaining energy levels throughout the day."),
    Document(page_content = "The efficiency of converting light into usable energy depends on several environmental factors."),
    Document(page_content = "Insufficient exposure to sunlight can significantly affect overall performance and growth."),
    Document(page_content = "The system relies heavily on sunlight to sustain its energy requirements."),
    Document(page_content = "Energy production decreases noticeably during periods of reduced sunlight."),
    Document(page_content = "Proper absorption of light is essential for optimal functioning and development."),
]   
```

```python
embedding_model = HuggingFaceEmbeddings(model_name = "sentence-transformers/all-MiniLM-L6-v2")
```

```python
llm = ChatGroq(model = "meta-llama/llama-4-scout-17b-16e-instruct")
```

```python
vector_Store = FAISS.from_documents(
    documents = docs, 
    embedding = embedding_model)
```

create MultQuery Retriever

```python
MultiQueryRetriever = MultiQueryRetriever.from_llm(
    llm = llm,
    retriever = vector_Store.as_retriever(search_kwargs = {"k": 3})
)
```

```python
query = "How to improve energy and maintain balance"
```

```python
results = MultiQueryRetriever.invoke(query)
```

```python
for i, doc in enumerate(results):
    print(f"Result {i+1}: {doc.page_content}\n")
```
