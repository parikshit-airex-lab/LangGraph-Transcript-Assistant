```python
from langchain_community.vectorstores import FAISS
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
```

Create source documents

```python
documents = [
    Document(page_content="Maharashtra is a state in the western peninsular region of India occupying a substantial portion of the Deccan plateau and extending to the Arabian Sea."),
    Document(page_content="Gujarat is a state on the western coast of India with a coastline of 1,600 km, most of which lies on the Kathiawar peninsula."),
    Document(page_content="Maharashtra is the second-most populous state in India and third-largest by area, while Gujarat is the fifth-largest state by area and ninth-largest by population."),
    Document(page_content="Madhya Pradesh is a state in central India. It is the second-largest state by area and the fifth-largest by population in India."),
    Document(page_content="Rajasthan is a state in northern India. It is the largest Indian state by area and the seventh-largest by population.")
]

```

```python
embedding_function=HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
vectorStore = FAISS.from_documents(
    documents = documents,
    embedding =embedding_function
)
```

convert vectorstore into retriver

```python
retriever = vectorStore.as_retriever(search_kwargs={"k":2})
```

Create query

```python
query1 = "Details of Gujrat in India"
```

```python
results = retriever.invoke(query1)
```
