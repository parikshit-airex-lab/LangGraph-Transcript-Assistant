```python
#from langchain_community.retrievers import MaximalMarginalRetriever
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS, Chroma
```

```python
docs = [ 
    Document(page_content="Langchain is a framework for building applications with LLMs. It provides a standard interface for all LLMs, as well as tools and utilities for working with them."),
    Document(page_content="ArithmeticError is the base class for those built-in exceptions that are raised for various arithmetic errors, such as division by zero, numeric overflow, and so on."),
    Document(page_content="Python is a high-level, interpreted programming language that is widely used for web development, data analysis, artificial intelligence, and scientific computing."),
    Document(page_content="Langchain provides a standard interface for all LLMs, as well as tools and utilities for working with them. It also provides a standard interface for all LLMs, as well as tools and utilities for working with them."),
    Document(page_content="Python is a high-level, interpreted programming language that is widely used for web development, data analysis, artificial intelligence, and scientific computing.")
]
```

```python
embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
vector_Store = Chroma.from_documents(
    documents = docs,
    embedding = embedding_model,
    collection_name = "langchain_docs"
)
```

Create retriever

```python
similarity_retriever = vector_Store.as_retriever(
    search_type = "similarity",
    search_kwargs = {"k":3,}
)
```

```python
mmr_retriever = vector_Store.as_retriever(
    search_type = "mmr",
    search_kwargs = {"k":3, "lambda_mult": 0.6}
)
```

```python
query = "What is langchain?"
```

```python
results = mmr_retriever.invoke(query)
results2 = similarity_retriever.invoke(query)
```

```python
for i, doc in enumerate(results):
    print(f"Result {i+1}: {doc.page_content}\n")
```
