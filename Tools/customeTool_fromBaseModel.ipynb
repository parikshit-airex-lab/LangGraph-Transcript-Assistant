```python
from langchain_core.tools import BaseTool
from pydantic import BaseModel, Field
from typing import Type
```

Create BaseModel for Schema

```python
class MultiplyInput(BaseModel):
    a: int = Field(required=True, description="First Integer Number")
    b: int = Field(required=True, description="Second Integer Number")
```

Create Tool Class

```python
class MultiplyTool(BaseTool):
    name : str = "Multiply",
    description : str = "Multiply two numbers",
    args_schema : Type[BaseModel] = MultiplyInput
    def _run(self, a:int, b:int) -> int:
        return a*b
```

create object of tool class

```python
Multiply_tool = MultiplyTool()
```

```python
result = Multiply_tool.invoke({"a":13, "b":5})
print(result)
```

```python
print(Multiply_tool.name)
print(Multiply_tool.description)
print(Multiply_tool.args_schema.model_json_schema())
```
