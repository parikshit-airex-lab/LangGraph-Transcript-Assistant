```python
from langchain_core.tools import tool
```

Step 1 - Create Function with type hints and add @tool decorator at top

```python
@tool
def multiply(a:int, b:int) -> int:
    """Multiply the two numbers """
    return a*b
```

Invoke tool

```python
result = multiply.invoke({"a":13, "b":5})
print(result)
print(multiply.name)
print(multiply.description)
print(multiply.args)
print(multiply.args_schema.model_json_schema())
```
